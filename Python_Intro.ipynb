```python
# if-elif-else ->used to check multiple condition
mark=int(input("Enter the student mark"))
if mark>=95:
    print("Grade A")
elif mark>=75:
    print("Grage B")
elif mark>=50:
    print("Grade C")
else:
    print("Grade F")
```
